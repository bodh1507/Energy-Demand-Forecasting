```python
uploaded = files.upload()
df1 = pd.read_csv("india_monthly_full_release_long_format.csv")
display(df1.head())
```

```python
df1 = df1[df1['Subcategory'] != 'Aggregate fuel']
df1 = df1.iloc[:, :-2]
display(df1.head())
```

```python
df1['Date'] = pd.to_datetime(df1['Date'])
df1['Year'] = df1['Date'].dt.year
df1['Month'] = df1['Date'].dt.month
df1 = df1.drop(columns=['Date'])

# Get current column names
cols = df1.columns.tolist()

# Remove 'Year' and 'Month' from their current positions
cols.remove('Year')
cols.remove('Month')

# Insert 'Year' and 'Month' at the beginning of the list
new_cols_order = ['Year', 'Month'] + cols

# Reindex the DataFrame with the new column order
df1 = df1[new_cols_order]
display(df1.head())
```

```python
cat = df1['Category'].unique()
print(cat)
df1_capacity = df1[df1['Category'] == 'Capacity']
display(df1_capacity.head())

df1_electricity_generation = df1[df1['Category'] == 'Electricity generation']
display(df1_electricity_generation.head())

df1_power_sector_emissions = df1[df1['Category'] == 'Power sector emissions']
display(df1_power_sector_emissions.head())
```

```python
df1_electricity_generation_monthly_state = df1_electricity_generation.groupby(['Year', 'Month', 'State'])['Value'].sum().reset_index()
display(df1_electricity_generation_monthly_state.head())
```

```python
df1_electricity_generation_yearly = df1_electricity_generation.groupby(['Year'])['Value'].mean().reset_index()
display(df1_electricity_generation_yearly.head())
```

```python
df1_electricity_generation_monthly_state.to_csv('electricity_gener_yearly.csv', index=False)
files.download('electricity_gener_yearly.csv')
```

```python
uploaded = files.upload()
df2 = pd.read_excel("temperature.xlsx")
display(df2.head())
```

```python
# Drop the initial descriptive row from df2 before any filtering
df2_cleaned = df2.iloc[1:].copy()

# Convert 'Year' column to numeric, coercing errors to NaN
df2_cleaned['Year'] = pd.to_numeric(df2_cleaned['Year'], errors='coerce')

# Filter for rows where 'Year' is not NaN and is within the desired range (2019-2024)
df2_filtered = df2_cleaned[df2_cleaned['Year'].notna() &
                         (df2_cleaned['Year'] >= 2019) &
                         (df2_cleaned['Year'] <= 2024)].copy()

# Convert 'Year' column to integer type
df2_filtered['Year'] = df2_filtered['Year'].astype(int)

# Define the monthly temperature columns
month_cols = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# Melt the DataFrame to transform monthly columns into rows
temp_melted = df2_filtered.melt(id_vars=['Year'], value_vars=month_cols, var_name='Month_Name', value_name='Temperature')

# Map month names to numerical month values
month_name_to_num = {
    'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4, 'May': 5, 'Jun': 6,
    'Jul': 7, 'Aug': 8, 'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec': 12
}
temp_melted['Month'] = temp_melted['Month_Name'].map(month_name_to_num)

# Select relevant columns for merging
temp_final = temp_melted[['Year', 'Month', 'Temperature']].copy()

# Merge df1_electricity_generation_monthly_state with temp_final
df_merged = pd.merge(df1_electricity_generation_monthly_state,
                     temp_final,
                     on=['Year', 'Month'],
                     how='left')

# Display the first few rows of the merged DataFrame
display(df_merged.head())
```

```python
df_merged.to_csv('merged_electricity_temperature_data.csv', index=False)
from google.colab import files
files.download('merged_electricity_temperature_data.csv')
```

```python
uploaded = files.upload()
df3 = pd.read_excel("gdp.xls")
display(df3.head())
```

```python
# Load df3 again, specifying header=2 to correctly interpret column names
df3_gdp = pd.read_excel("gdp.xls", header=3)

# Filter for India and the specific indicator 'GDP (constant 2015 US$)'
india_gdp = df3_gdp[ (df3_gdp['Country Name'] == 'India') &
                     (df3_gdp['Indicator Name'] == 'GDP (constant 2015 US$)')
                   ].copy()

# Select only the relevant year columns and 'Country Name'
years = [str(year) for year in range(2019, 2025)]
india_gdp = india_gdp[['Country Name'] + years]

# Melt the DataFrame to transform year columns into rows
india_gdp_melted = india_gdp.melt(id_vars=['Country Name'],
                                 var_name='Year',
                                 value_name='GDP')

# Convert 'Year' to integer and 'GDP' to numeric
india_gdp_melted['Year'] = pd.to_numeric(india_gdp_melted['Year'], errors='coerce').astype(int)
india_gdp_melted['GDP'] = pd.to_numeric(india_gdp_melted['GDP'], errors='coerce')

# Select relevant columns for merging
gdp_final = india_gdp_melted[['Year', 'GDP']].copy()

# Merge df_merged with gdp_final on 'Year'
df_merged_with_gdp = pd.merge(df_merged,
                              gdp_final,
                              on='Year',
                              how='left')

# Display the first few rows of the updated merged DataFrame
display(df_merged_with_gdp.head())
```

```python
df_merged_with_gdp.to_csv('final_merged.csv', index=False)
from google.colab import files
files.download('final_merged.csv')
```
